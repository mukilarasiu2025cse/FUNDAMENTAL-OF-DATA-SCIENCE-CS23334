```python
import seaborn as sns
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

%matplotlib inline

# Load the tips dataset
tips = sns.load_dataset('tips')

# 1. Display the first five rows
print("OUTPUT 1: FIRST FIVE ROWS")
display(tips.head())

# 2. Distribution plot with KDE
sns.displot(data=tips, x='total_bill', kde=True)
plt.title("Total Bill Distribution with KDE")
plt.show()

# 3. Distribution plot without KDE
sns.displot(data=tips, x='total_bill', kde=False)
plt.title("Total Bill Distribution")
plt.show()

# 4. Joint plot
sns.jointplot(data=tips, x='tip', y='total_bill')
plt.show()

# 5. Joint plot with regression line
sns.jointplot(data=tips, x='tip', y='total_bill', kind='reg')
plt.show()

# 6. Hexagonal joint plot
sns.jointplot(data=tips, x='tip', y='total_bill', kind='hex')
plt.show()

# 7. Pair plot
sns.pairplot(tips)
plt.show()

# 8. Count values of time
print("OUTPUT 8: TIME VALUE COUNTS")
display(tips.time.value_counts())

# 9. Pair plot grouped by time
sns.pairplot(tips, hue='time')
plt.show()

# 10. Pair plot grouped by day
sns.pairplot(tips, hue='day')
plt.show()

# 11. Heatmap of numerical correlations
plt.figure(figsize=(8, 6))
sns.heatmap(
    tips.corr(numeric_only=True),
    annot=True,
    cmap='coolwarm'
)
plt.title("Correlation Heatmap")
plt.show()

# 12. Box plot of total bill
sns.boxplot(x=tips.total_bill)
plt.title("Box Plot of Total Bill")
plt.show()

# 13. Box plot of tip
sns.boxplot(x=tips.tip)
plt.title("Box Plot of Tip")
plt.show()

# 14. Count plot of day
sns.countplot(data=tips, x='day')
plt.title("Count of Customers by Day")
plt.show()

# 15. Count plot of sex
sns.countplot(data=tips, x='sex')
plt.title("Count by Sex")
plt.show()

# 16. Pie chart of sex
tips.sex.value_counts().plot(kind='pie', autopct='%1.1f%%')
plt.title("Customer Distribution by Sex")
plt.ylabel("")
plt.show()

# 17. Bar chart of sex
tips.sex.value_counts().plot(kind='bar')
plt.title("Customer Count by Sex")
plt.xlabel("Sex")
plt.ylabel("Count")
plt.show()

# 18. Count plot of days for Dinner
sns.countplot(data=tips[tips.time == 'Dinner'], x='day')
plt.title("Dinner Customers by Day")
plt.show()

print("EXPERIMENT NO. 15 COMPLETED SUCCESSFULLY")
```
