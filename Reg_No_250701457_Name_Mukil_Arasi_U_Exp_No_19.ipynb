```python
import numpy as np
import pandas as pd

# Load the dataset
url = "https://gist.githubusercontent.com/davewadestein/5339c464e95f54f724f6f5381f0d95e1/raw/"

try:
    df = pd.read_csv(url)
    if 'Purchased' not in df.columns:
        raise ValueError("Dataset columns not found")
except Exception:
    print("Download failed. Please upload Social_Network_Ads.csv.")
    from google.colab import files
    uploaded = files.upload()
    filename = next(iter(uploaded))
    df = pd.read_csv(filename)

# Display dataset
print(df.head())
df.info()

# Select features and label
features = df.iloc[:, [2, 3]].values
label = df.iloc[:, 4].values

print(features)
print(label)

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report, confusion_matrix

# Find suitable random states
for i in range(1, 401):
    x_train, x_test, y_train, y_test = train_test_split(
        features, label, test_size=0.2, random_state=i
    )

    model = LogisticRegression(max_iter=1000)
    model.fit(x_train, y_train)

    train_score = model.score(x_train, y_train)
    test_score = model.score(x_test, y_test)

    if test_score > train_score:
        print("Test {} Train{} Random State {}".format(
            test_score, train_score, i
        ))

# Train the final model
x_train, x_test, y_train, y_test = train_test_split(
    features, label, test_size=0.2, random_state=26
)

finalModel = LogisticRegression(max_iter=1000)
finalModel.fit(x_train, y_train)

# Display accuracy
print(finalModel.score(x_train, y_train))
print(finalModel.score(x_test, y_test))

# Classification report
print(classification_report(label, finalModel.predict(features)))

# Confusion matrix for test data
print(confusion_matrix(y_test, finalModel.predict(x_test)))
```
