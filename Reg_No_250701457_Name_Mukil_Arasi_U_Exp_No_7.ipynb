```python
import numpy as np
import matplotlib.pyplot as plt

# Define population parameters
population_mean = 75
population_std = 12
population_size = 40000

# Generate a normally distributed population
population = np.random.normal(population_mean, population_std, population_size)
sample_sizes = [15, 40, 80]
num_samples = 300

# Dictionary to store sample means for each sample size
sample_means = {}

# Simulate sampling distributions
for size in sample_sizes:
    sample_means[size] = []
    for _ in range(num_samples):
        sample = np.random.choice(population, size=size, replace=False)
        sample_means[size].append(np.mean(sample))

# Plotting the results
plt.figure(figsize=(12, 8))

for i, size in enumerate(sample_sizes):
    plt.subplot(len(sample_sizes), 1, i + 1)
    plt.hist(sample_means[size], bins=25, alpha=0.7, label=f'Sample Size {size}')

    # Draw a vertical line for the true population mean
    plt.axvline(np.mean(population), color='red', linestyle='dashed', linewidth=1.5, label='Population Mean')

    plt.title(f'Sampling Distribution (Sample Size {size})')
    plt.xlabel('Sample Mean')
    plt.ylabel('Frequency')
    plt.legend()

plt.tight_layout()
plt.show()
```
