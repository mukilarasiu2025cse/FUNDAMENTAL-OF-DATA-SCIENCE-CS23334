```python
import numpy as np
import scipy.stats as stats

# Step 1: Define sample bird weights in grams
sample_data = np.array([
    152, 148, 151, 149, 147, 153, 150, 148, 152, 149,
    151, 150, 149, 152, 151, 148, 150, 152, 149, 150,
    148, 153, 151, 150, 149, 152, 148, 151, 150, 153
])

# Step 2: Set the hypothesized population mean
population_mean = 150

# Step 3: Calculate sample statistics
sample_mean = np.mean(sample_data)
sample_std = np.std(sample_data, ddof=1)
n = len(sample_data)

# Step 4: Calculate the Z-statistic
z_statistic = (
    (sample_mean - population_mean) /
    (sample_std / np.sqrt(n))
)

# Step 5: Calculate the two-tailed p-value
p_value = 2 * stats.norm.sf(abs(z_statistic))

# Step 6: Set significance level
alpha = 0.05

# Step 7: Display results
print("Sample Size:", n)
print(f"Sample Mean: {sample_mean:.2f}")
print(f"Sample Standard Deviation: {sample_std:.4f}")
print(f"Z-Statistic: {z_statistic:.4f}")
print(f"P-Value: {p_value:.4f}")
print(f"Significance Level: {alpha}")

# Step 8: Make a decision
if p_value < alpha:
    print("Reject the null hypothesis.")
    print("The average bird weight is significantly different from 150 grams.")
else:
    print("Fail to reject the null hypothesis.")
    print("There is insufficient evidence that the average bird weight differs from 150 grams.")

```
